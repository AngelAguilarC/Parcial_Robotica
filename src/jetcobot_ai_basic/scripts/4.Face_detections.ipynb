```python
import cv2 as cv
import threading
from time import sleep
import ipywidgets as widgets
from IPython.display import display
from face_follow import face_follow
from jetcobot_utils.grasp_controller import GraspController
```

```python
graspController = GraspController()
follow = face_follow()
# 初始化机械臂位姿 Initialize the robot arm pose
graspController.init_pose()
model = 'General'
```

```python
# 图像控件 Image widget
imgbox = widgets.Image(format='jpg', height=480, width=640, layout=widgets.Layout(align_self='center'))
```

```python
capture = cv.VideoCapture(0)
```

```python

def camera():
    global model
    # 打开摄像头 Open camera
    while capture.isOpened():
        _, img = capture.read()
        img = cv.resize(img, (640, 480))
        img = follow.follow_function(img)
        if model == 'Exit':
            break
        imgbox.value = cv.imencode('.jpg', img)[1].tobytes()
```

```python
display(imgbox)

try:
    camera()
except:
    pass
```

```python
capture.release()
```
