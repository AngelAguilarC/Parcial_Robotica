```python
import cv2
import ipywidgets.widgets as widgets
import threading
import time
```

```python
#bgr8转jpeg格式
def bgr8_to_jpeg(value, quality=75):
    return bytes(cv2.imencode('.jpg', value)[1])
```

```python
g_camera = cv2.VideoCapture(0,cv2.CAP_V4L2)

g_camera.set(cv2.CAP_PROP_FRAME_WIDTH, 640)#设置图像宽度 Set the image width
g_camera.set(cv2.CAP_PROP_FRAME_HEIGHT,480)#设置图像高度 Set the image height
g_camera.set(cv2.CAP_PROP_FOURCC, cv2.VideoWriter.fourcc('M', 'J', 'P', 'G'))

ret, frame = g_camera.read()     #读取摄像头数据 Reading camera data

image_widget = widgets.Image(format='jpeg', width=640, height=480)  #设置摄像头显示组件 Set up the camera display component
image_widget.value = bgr8_to_jpeg(frame)
```

```python
display(image_widget)      #显示摄像头组件 Display camera components
try:
    m_fps = 0
    t_start = time.time()
    while g_camera.isOpened():
        ret, frame = g_camera.read()
        if not ret:
            print("Camera Read Fail")
            break

        m_fps = m_fps + 1
        fps = m_fps / (time.time() - t_start)
        if (time.time() - t_start) >= 2:
            m_fps = fps
            t_start = time.time() - 1

        text="FPS:" + str(int(fps))
        cv2.putText(frame, text, (10, 20), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 255), 1)
        image_widget.value = bgr8_to_jpeg(frame)
except:
    print(" Program closed! ")
    pass
```

```python
g_camera.release()              #使用完成对象记住释放掉对象，不然下一个程序使用这个对象模块会被占用，导致无法使用 After using the object, remember to release the object, otherwise the object module will be occupied by the next program, making it unusable
```
